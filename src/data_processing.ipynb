```python
# import libraries
import pandas as pd
```

```python
# load csv files 
df1 = pd.read_csv('../data/daily_sales_data_0.csv')
df2 = pd.read_csv('../data/daily_sales_data_1.csv')
df3 = pd.read_csv('../data/daily_sales_data_2.csv')
```

```python
# read the dataframes
print(df1.head())
print(df2.head())
print(df3.head())
```

```python
# concatenate the dataframes
df = pd.concat([df1, df2, df3])
```

```python
# Keep only Pink Morsel Product

df = df[df['product'] == 'pink morsel']
```

```python
# remove $ sign from price column 
df['price'] = (df['price'].replace("[$]", "", regex=True).astype(float))
```

```python
# Create Sales Column by multiplying quantity and price
df['sales'] = df['quantity'] * df['price']
```

```python
df.head()
```

```python
# keep only the required columns
final_df = df[["sales", "date", "region"]]
```

```python
# Rename the columns
final_df.columns = ['sales', 'date', 'region']
```

```python
# Export final dataframe to csv
final_df.to_csv('../output/pink_morsel_sales.csv', index=False)
```

```python
print("Data processing complete. Final data saved to '../output/pink_morsel_sales.csv'")
```
